## Selected Lane: Refresh / Content Opportunity Scoring (Core Lane)

Why: In an enterprise environment, human editorial time is a strictly constrained resource. The goal isn't just to build an isolated predictive model; the goal is to build an automated prioritization engine to direct limited resources to the content that will yield the highest ROI upon refresh.

## The question: decision, action, cost of a wrong call
The Decision: Which specific URLs should the content and SEO teams prioritize for immediate updates this week?

The Action: A human editor is assigned to rewrite or update the flagged URL.

The Cost of a Wrong Call: Wasting expensive, limited human editorial hours on a page that is actually stable, while high-value pages are left to bleed traffic.

Why ML helps: The pattern of traffic decay is real but too messy to write by hand. It involves many tangled, shifting signals (position tier, CTR degradation, and historical impressions) that a simple IF/THEN rule cannot capture.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset directly from your public GitHub URL instead of a local folder
url = "https://raw.githubusercontent.com/MaazKhan53/ML-01-Run-the-Starter-Notebooks/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Metric 1: The baseline rate of decay
declining_rate = (df["trend_direction"] == "down").mean()

# Metric 2: The traffic volume at stake for these decaying pages
declining_pages = df[df["trend_direction"] == "down"]
median_impressions_down = declining_pages["impressions_90d"].median()

# Print the real numbers to prove the lane is worth it
print(f"1. Baseline Decay Rate: {declining_rate:.1%} of the pages are actively declining.")
print(f"2. Volume at Stake: The median declining page has {median_impressions_down:,.0f} impressions over 90 days.")
```
