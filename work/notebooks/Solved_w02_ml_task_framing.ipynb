## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

# Load the starter dataset directly from your public GitHub URL
url = "https://raw.githubusercontent.com/MaazKhan53/ML-01-Run-the-Starter-Notebooks/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# ENGINEER THE TARGET: Create the binary label from the raw trend_direction text
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)

# The unit of analysis: One row = one content item (page/URL) for a specific client.
# Isolate the identifiers, pre-decision features, and the newly created target column.
unit_df = df[['client_id', 'content_id', 'content_age_days', 'avg_position', 'is_declining_label']]

print(f"Dataset size: {len(unit_df):,} rows.")
print("Unit of analysis: One row = One specific piece of content (URL).")
print("Target sketched: 'is_declining_label' (1 = declining, 0 = stable/growing).")
print("-" * 50)

# Display the real dataframe slice
unit_df.head()
```

## Why ML wins:
As demonstrated in the baseline audits, a human-written rule (e.g., stale * visible) works reasonably well at the absolute top of the list but runs out of signal quickly. Traffic decay is a tangled interaction of multiple shifting signals—such as click-through-rate cliffs, position tier drops, and engagement rates. ML earns its place here because it can weigh these multivariate signals simultaneously to uncover hidden patterns of decay that simple IF/THEN business logic misses.
